```python

import pandas as pd

# Load the dataset into Pandas
df = pd.read_csv('bayut_selling_properties.csv')

# Check total rows and column names
print("--- SUCCESS: RAW DATA LOADED ---")
print(f"Total Rows: {df.shape[0]:,}")
print(f"Total Columns: {df.shape[1]}")
print("\nAvailable Columns:")
print(df.columns.tolist())
```

```python

# 1. Count total duplicate rows exist across the entire dataset
duplicate_count = df.duplicated().sum()
print(f"Number of exact duplicate rows found: {duplicate_count}")

# 2. Drop the duplicates and save the cleaned data back into 'df'
df = df.drop_duplicates()

# 3. Print the new row count to verify they are gone
print(f"Total rows remaining after removing duplicates: {df.shape[0]:,}")
```

```python
# 1. Identify rows where critical property details are identical
# We will check if 'price', 'type', 'beds', and 'baths' match exactly
smart_duplicates = df.duplicated(subset=['price', 'type', 'beds', 'baths']).sum()
print(f"Number of duplicate listings based on property details: {smart_duplicates}")

# 2. Drop these partial duplicates, keeping only the first one
df = df.drop_duplicates(subset=['price', 'type', 'beds', 'baths'], keep='first')

# 3. Print the final cleaned row count
print(f"Total rows remaining now: {df.shape[0]:,}")
```

```python
# 1. Calculate total missing values in each column
missing_data = df.isnull().sum()

# 2. Filter to show only the columns that actually have missing values
missing_only = missing_data[missing_data > 0]

print("COLUMNS WITH MISSING VALUES")
if len(missing_only) == 0:
    print("No missing values found!")
else:
    print(missing_only)
```

```python
# 1. Drop rows where Latitude or Longitude are missing
df = df.dropna(subset=['Latitude', 'Longitude'])

# 2. Verify that there are absolutely zero missing values left
print("Remaining missing values count:")
print(df[['Latitude', 'Longitude']].isnull().sum())

# 3. Check the final clean row count
print(f"\nTotal clean rows now: {df.shape[0]:,}")
```

```python
# 1. Check the current data type of the price column
print(f"Price column data type: {df['price'].dtype}")

# 2. Checking first 5 raw values to see if they contain formatting text/commas
print("\nFirst 5 values in price:")
print(df['price'].head())
```

```python
# 1. Print descriptive statistics for the price column
print("PRICE STATISTICS")
print(f"Minimum Price: {df['price'].min():,}")
print(f"Maximum Price: {df['price'].max():,}")
print(f"Average Price: {df['price'].mean():,.2f}")
```

```python
# 1. Count how many properties have a price of 0 aed
zero_prices = (df['price'] == 0).sum()
print(f"Number of properties listed with 0 AED: {zero_prices}")

# 2. Keep only properties where price is greater than 0
df = df[df['price'] > 0]

# 3. Print out the updated minimum price to verify
print(f"New Minimum Price: {df['price'].min():,}")
print(f"Total rows remaining: {df.shape[0]:,}")
```

```python
# 1. Print unique values for text-based columns
print("UNIQUE VALUES IN PURPOSE")
print(df['purpose'].unique())

print("\nUNIQUE VALUES IN TYPE")
print(df['type'].unique())
```

```python
# 1. Export the cleaned DataFrame to a new CSV file
df.to_csv('cleaned_uae_properties.csv', index=False)

# 2. Print a final summary to celebrate your data cleaning project milestones
print(" SUCCESS: CLEANED DATA EXPORTED COMPLETE! ")
print(f"Final Row Count: {df.shape[0]:,}")
print(f"Final Column Count: {df.shape[1]}")
print("Your clean file 'cleaned_uae_properties.csv' is ready in your sidebar!")
```
